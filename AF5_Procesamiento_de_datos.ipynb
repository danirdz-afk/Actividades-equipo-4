```python
print(df.isnull().sum())
```

```python
# Filtra velocidad si es negativa
df_velocidad_negativa = df[df['Velocidad (cm/s)'] < 0]
print(df_velocidad_negativa)
```

```python
#Normalizar los datos numéricos con Min-Max Scaling
print(F"*******************DISTANCIA**************************")
distancia_lejana = df.loc[df['Distancia (cm)'].idxmax(), 'Distancia (cm)']
distancia_corta = df.loc[df['Distancia (cm)'].idxmin(), 'Distancia (cm)']
print(f"distancia lejana: {distancia_lejana}")
print(f"distancia corta: {distancia_corta}")
print(F"********************VELOCIDAD*************************")
velocidad_alta = df.loc[df['Velocidad (cm/s)'].idxmax(), 'Velocidad (cm/s)']
velocidad_baja = df.loc[df['Velocidad (cm/s)'].idxmin(), 'Velocidad (cm/s)']
print(f"Velocidad alta: {velocidad_alta}")
print(f"Velocidad baja: {velocidad_baja}")
print(F"*******************TIEMPO**************************")
tiempo_alto = df.loc[df['Tiempo (s)'].idxmax(), 'Tiempo (s)']
tiempo_bajo = df.loc[df['Tiempo (s)'].idxmin(), 'Tiempo (s)']
print(f"Tiempo alto: {tiempo_alto}")
print(f"Tiempo bajo: {tiempo_bajo}")
print(F"********************TEMPERATURA*************************")
temperatura_alta = df.loc[df['Temperatura (°C)'].idxmax(), 'Temperatura (°C)']
temperatura_baja = df.loc[df['Temperatura (°C)'].idxmin(), 'Temperatura (°C)']
print(f"Temperatura alta: {temperatura_alta}")
print(f"Temperatura baja: {temperatura_baja}")
```
